## HNSW

#### Create embeddings

````python
text_content = [doc.page_content for doc in chunks]
vectors = embeddings.embed_documents(text_content)

````

#### step 2 : Convert to NumPy


````python
import numpy as np
vectors = np.array(vectors).astype("float32")
````

#### Step 3: Create HNSW

````python
import faiss
dimension = vectors.shape[1]
````

````python
vectors.shape[1]
````

````python
index = faiss.IndexHNSWFlat(
    dimension,
    32
)
````

#### Step 4 : Add Vectors

````python
index.add(vectors)
#now you are using HNSW
````

#### Step 5 : Create Docstore

````python
from langchain_community.docstore.in_memory import InMemoryDocstore
````

````python
docstore = InMemoryDocstore()
index_to_docs_store_id = {}
for i, doc in enumerate(chunks):
    doc_id = str(i)
    docstore.add(
        {
            doc_id:doc
        }
    )
    index_to_docs_store_id[i] = doc_id
````

#### Step 6: Creating FAISS HNSW Vector Database (and save it)


````python
from langchain_community.vectorstores import FAISS
````

````python
vectorstore = FAISS(
    embedding_function=embeddings,
    index=index,
    docstore=docstore,
    index_to_docstore_id=index_to_docs_store_id
)
````

#### Step 7: Create Retriever

````python
retriever = vectorstore.as_retriever(
    search_kwargs = {
        "k":3
    }
)
````

#### step 8: ask question 

````python
docs = retriever.invoke(
    "what is vibecoding"
)
````

````python
for doc in docs:
    print(doc.page_content)
````
