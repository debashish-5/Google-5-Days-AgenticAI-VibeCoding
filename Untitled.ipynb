```python
vectorstore.save_local("faiss_index")

```

* `index.faiss`: The actual FAISS index containing your vectors.
* `index.pkl`: The pickled document store and index-to-ID mapping required to retrieve your original text.

When you want to reload this in a new session, you simply point to the folder and pass your `embeddings` object. You **do not** need to redefine `index`, `docstore`, or `index_to_docs_store_id`:

```python
from langchain_community.vectorstores import FAISS

# This will automatically reconstruct the index, docstore, and mappings
new_vectorstore = FAISS.load_local(
    "my_faiss_index", 
    embeddings, 
    allow_dangerous_deserialization=True
)

```

### Why this is easier:

* **Encapsulation:** The `save_local` method bundles the FAISS index (`.faiss`) and the document store (`.pkl`) into the directory you specify.
* **Reconstruction:** `load_local` reads those files and maps them back into the `FAISS` class structure for you.
* **Consistency:** As long as you provide the same `embedding_function` that you used when creating the index, everything will link back together correctly.

class Retriever:
    def __init__(self, llm=None, embeddings=None):
        # 1. Fallback to Default LLM if none is provided manually
        if llm is None:
            self.llm = ChatGroq(
                model="llama-3.1-8b-instant", 
                api_key=os.getenv("GROQ_API_KEY")
            )
        else:
            self.llm = llm
            
        # 2. Fallback to Default Embeddings if none is provided manually
        if embeddings is None:
            self.embeddings = HuggingFaceEmbeddings(
                model_name="sentence-transformers/all-MiniLM-L6-v2"
            )
        else:
            self.embeddings = embeddings
            
        # 3. Load your Local Vector Store using the chosen embeddings
        self.vector_store = FAISS.load_local(
            folder_path="vector_store", 
            embeddings=self.embeddings,
            allow_dangerous_deserialization=True
        )

    def get_retriever(self):
        return self.vector_store.as_retriever(search_kwargs={"k": 5})
